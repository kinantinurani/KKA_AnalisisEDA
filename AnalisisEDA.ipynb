```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Pada NumPy array, harga * 0.9 akan mengalikan setiap elemen dengan 0,9.

Sedangkan pada list Python, [5000,7000,3000] * 0.9 menghasilkan TypeError karena list hanya bisa dikalikan dengan bilangan bulat untuk mengulang isi list.

Kolom yang kosong adalah menu dan terjual. Jika tidak dibersihkan, data kosong dapat membuat hasil analisis atau perhitungan menjadi tidak akurat.

```python
df = pd.read_csv('dataset_penjualan_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris total berarti kolom tersebut memiliki data kosong (missing value).

Artinya, ada beberapa baris yang belum memiliki nilai pada kolom tersebut sehingga perlu dibersihkan sebelum dianalisis.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Karena terjual yang kosong masih bisa dianggap 0 transaksi, jadi diisi dengan 0.

Sedangkan menu adalah data penting untuk mengetahui produk apa yang dijual. Jika kosong, baris tersebut sulit digunakan dalam analisis, sehingga barisnya dihapus.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Setelah drop_duplicates(), jumlah baris berkurang sesuai jumlah data yang duplikat.

Memastikan dtypes benar penting agar perhitungan dan analisis data tidak salah, misalnya kolom harga harus bertipe angka agar bisa digunakan untuk operasi matematika.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Menu dengan total pendapatan tertinggi adalah menu yang memiliki nilai total_pendapatan paling besar pada hasil groupby.

Informasi ini membantu kantin untuk menambah stok dan memprioritaskan menu yang paling menguntungkan agar pendapatan meningkat.

```python
#Tugas Kelompok

import pandas as pd

# 1. DATA LOADING & INSPECTION
df = pd.read_csv('dataset_penjualan_kantin.csv')
print('=== 1. INSPEKSI DATASET AWAL ===')
print('Head (5 Baris Pertama):')
print(df.head())

print('Info Dataset:')
print(df.info())

print('Describe Dataset:')
print(df.describe(include='all'))

print('Shape (Jumlah Baris & Kolom):')
print(df.shape)

# 2. DATA CLEANING
df_clean = df.copy()
# A. Hapus Duplikat
df_clean = df_clean.drop_duplicates()

# B. Penyeragaman Teks Kategori
df_clean['kategori'] = df_clean['kategori'].str.capitalize()

# C. Menangani Missing Value (Fillna)
# Isi terjual yang NaN dengan 0
df_clean['terjual'] = df_clean['terjual'].fillna(0)

# Isi harga yang NaN berdasarkan median harga produk sejenis
harga_median = df_clean.groupby('menu')['harga'].median()
df_clean['harga'] = df_clean['harga'].fillna(
    df_clean['menu'].map(harga_median)
)
print('=== 2. PEMBERSIHAN DATA SELESAI ===')
print(df_clean.info())


# 3. DATA MANIPULATION
# A. Kolom Turunan
df_clean['total_pendapatan'] = (
    df_clean['harga'] * df_clean['terjual']
)

# B. Filtering (Transaksi terjual > 5)
transaksi_besar = df_clean[df_clean['terjual'] > 5]

# C. Sorting (Pendapatan Tertinggi)
df_sorted = df_clean.sort_values(by='total_pendapatan', ascending=False)

# D. Groupby / Agregasi (Pendapatan per Menu)
pendapatan_per_menu = (
    df_clean.groupby('menu')['total_pendapatan']
    .sum()
    .reset_index()
    .sort_values(by='total_pendapatan', ascending=False)
)

# Groupby per Kategori
pendapatan_per_kategori = (
    df_clean.groupby('kategori')['total_pendapatan']
    .sum()
    .reset_index()
    .sort_values(by='total_pendapatan', ascending=False)
)

print('=== 3. HASIL MANIPULASI DATA ===')
print('Pendapatan per Menu:')
print(pendapatan_per_menu)
print('Pendapatan per Kategori:')
print(pendapatan_per_kategori)

# 4. SIMPAN DATASET BERSIH
df_clean.to_csv('dataset_bersih.csv', index=False)
print("[SUCCESS] Dataset bersih berhasil disimpan dengan nama 'dataset_bersih.csv'!")

# 5. DATA PROFILING SUMMARY
"""
RINGKASAN TEMUAN UTAMA:
1. 'Roti Bakar' dan 'Nasi Goreng' merupakan kontributor pendapatan terbesar di kantin sekolah.
2. Kategori Makanan secara keseluruhan menyumbang proporsi pendapatan tertinggi dibandingkan Minuman dan Snack.
3. Setelah proses pembersihan data, data yang kosong dan terduplikasi berhasil ditangani dengan baik.
"""
```
